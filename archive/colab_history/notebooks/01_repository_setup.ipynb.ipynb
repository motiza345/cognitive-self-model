````python
import os
import stat
import tempfile
import subprocess
from pathlib import Path
from getpass import getpass
from urllib.parse import urlparse

PROJECT_ROOT = Path("/content/cognitive-self-model")

if not (PROJECT_ROOT / ".git").exists():
    raise RuntimeError("Git Repository معتبر پیدا نشد.")


# ============================================================
# 1. دریافت و بررسی آدرس Remote
# ============================================================

remote_url = subprocess.check_output(
    ["git", "remote", "get-url", "origin"],
    cwd=PROJECT_ROOT,
    text=True,
).strip()

parsed_remote = urlparse(remote_url)

if (
    parsed_remote.scheme != "https"
    or parsed_remote.hostname != "github.com"
):
    raise RuntimeError(
        "آدرس origin باید یک آدرس HTTPS معتبر از GitHub باشد.\n"
        f"Remote فعلی: {remote_url}"
    )

print("Remote:", remote_url)


# ============================================================
# 2. تشخیص Branch فعلی
# ============================================================

current_branch = subprocess.check_output(
    ["git", "branch", "--show-current"],
    cwd=PROJECT_ROOT,
    text=True,
).strip()

if not current_branch:
    raise RuntimeError("Branch فعلی قابل تشخیص نیست.")

print("Branch:", current_branch)


# ============================================================
# 3. دریافت امن اطلاعات ورود
# ============================================================

github_username = input(
    "GitHub username: "
).strip()

github_token = getpass(
    "GitHub Personal Access Token (مخفی می‌ماند): "
).strip()

if not github_username:
    raise ValueError("GitHub username وارد نشده است.")

if not github_token:
    raise ValueError("GitHub Token وارد نشده است.")


# ============================================================
# 4. ایجاد AskPass موقت
# Token داخل Remote یا Git config ذخیره نمی‌شود
# ============================================================

askpass_path = None

try:
    with tempfile.NamedTemporaryFile(
        mode="w",
        suffix=".sh",
        delete=False,
        encoding="utf-8",
    ) as askpass_file:

        askpass_file.write(
            '#!/bin/sh\n'
            'case "$1" in\n'
            '  *Username*) printf "%s\\n" "$GITHUB_USERNAME" ;;\n'
            '  *Password*) printf "%s\\n" "$GITHUB_TOKEN" ;;\n'
            '  *) printf "\\n" ;;\n'
            'esac\n'
        )

        askpass_path = askpass_file.name

    os.chmod(
        askpass_path,
        stat.S_IRUSR | stat.S_IWUSR | stat.S_IXUSR,
    )

    safe_environment = os.environ.copy()

    safe_environment["GIT_ASKPASS"] = askpass_path
    safe_environment["GIT_TERMINAL_PROMPT"] = "0"
    safe_environment["GITHUB_USERNAME"] = github_username
    safe_environment["GITHUB_TOKEN"] = github_token

    print("\nدر حال ارسال Commit به GitHub...\n")

    result = subprocess.run(
        [
            "git",
            "push",
            "--set-upstream",
            "origin",
            current_branch,
        ],
        cwd=PROJECT_ROOT,
        env=safe_environment,
        text=True,
        capture_output=True,
    )

finally:
    # حذف Token از متغیرهای Python
    github_token = None

    # حذف فایل موقت AskPass
    if askpass_path and os.path.exists(askpass_path):
        os.remove(askpass_path)


# ============================================================
# 5. نمایش نتیجه
# ============================================================

if result.stdout.strip():
    print(result.stdout.strip())

if result.stderr.strip():
    print(result.stderr.strip())

if result.returncode != 0:
    raise RuntimeError(
        "Push ناموفق بود. متن خطای بالا را بررسی کنید؛ "
        "Token را ارسال نکنید."
    )

print("\n✅ Commit با موفقیت به GitHub ارسال شد.")
print("✅ Token در Remote یا Git config ذخیره نشده است.")
````

````python
from pathlib import Path
import json
import re
from google.colab import files

# ---------------------------------------------------------
# 1. Upload source file
# ---------------------------------------------------------

print("=" * 72)
print("COGNITIVE SELF-MODEL — SOURCE IMPORT INSPECTION")
print("=" * 72)

print(
    "\nفایل اصلی پروژه را انتخاب کنید.\n"
    "فرمت مجاز: .ipynb یا .py\n"
)

uploaded = files.upload()

if len(uploaded) != 1:
    raise ValueError(
        f"دقیقاً یک فایل باید انتخاب شود؛ تعداد انتخاب‌شده: {len(uploaded)}"
    )

original_name, file_bytes = next(iter(uploaded.items()))
safe_name = Path(original_name).name
suffix = Path(safe_name).suffix.lower()

if suffix not in {".ipynb", ".py"}:
    raise ValueError(
        f"فرمت فایل پشتیبانی نمی‌شود: {suffix}\n"
        "فقط فایل‌های .ipynb و .py مجاز هستند."
    )

# فایل خارج از Git Repository نگهداری می‌شود.
INCOMING_DIR = Path("/content/incoming_source")
INCOMING_DIR.mkdir(parents=True, exist_ok=True)

source_path = INCOMING_DIR / safe_name
source_path.write_bytes(file_bytes)

print(f"\n✅ فایل دریافت شد: {safe_name}")
print(f"مسیر موقت: {source_path}")
print(f"حجم فایل: {len(file_bytes):,} bytes")


# ---------------------------------------------------------
# 2. Extract source code without modifying it
# ---------------------------------------------------------

notebook_info = None

if suffix == ".ipynb":
    try:
        notebook = json.loads(file_bytes.decode("utf-8"))
    except UnicodeDecodeError as error:
        raise ValueError("Notebook دارای کدگذاری UTF-8 معتبر نیست.") from error
    except json.JSONDecodeError as error:
        raise ValueError("ساختار JSON فایل Notebook معتبر نیست.") from error

    cells = notebook.get("cells", [])

    code_cells = [
        cell for cell in cells
        if cell.get("cell_type") == "code"
    ]

    markdown_cells = [
        cell for cell in cells
        if cell.get("cell_type") == "markdown"
    ]

    raw_cells = [
        cell for cell in cells
        if cell.get("cell_type") == "raw"
    ]

    code_parts = []

    for index, cell in enumerate(code_cells, start=1):
        source = cell.get("source", [])

        if isinstance(source, list):
            source = "".join(source)

        code_parts.append(
            f"# {'=' * 68}\n"
            f"# ORIGINAL NOTEBOOK CODE CELL {index}\n"
            f"# {'=' * 68}\n"
            f"{source.rstrip()}\n"
        )

    extracted_code = "\n\n".join(code_parts)

    output_count = sum(
        len(cell.get("outputs", []))
        for cell in code_cells
    )

    notebook_info = {
        "total_cells": len(cells),
        "code_cells": len(code_cells),
        "markdown_cells": len(markdown_cells),
        "raw_cells": len(raw_cells),
        "stored_outputs": output_count,
    }

else:
    try:
        extracted_code = file_bytes.decode("utf-8")
    except UnicodeDecodeError as error:
        raise ValueError("فایل Python دارای کدگذاری UTF-8 معتبر نیست.") from error


# ---------------------------------------------------------
# 3. Static inspection
# ---------------------------------------------------------

code_lines = extracted_code.splitlines()

colab_directives = []
absolute_paths = []
possible_secret_locations = []

secret_patterns = {
    "GitHub token": re.compile(
        r"(?:ghp_[A-Za-z0-9]{20,}|github_pat_[A-Za-z0-9_]{20,})"
    ),
    "Hugging Face token": re.compile(
        r"\bhf_[A-Za-z0-9]{20,}\b"
    ),
    "OpenAI-style key": re.compile(
        r"\bsk-[A-Za-z0-9_-]{20,}\b"
    ),
    "Private key": re.compile(
        r"-----BEGIN (?:RSA |EC |OPENSSH )?PRIVATE KEY-----"
    ),
}

for line_number, line in enumerate(code_lines, start=1):
    stripped = line.lstrip()

    if stripped.startswith(("!", "%")):
        colab_directives.append((line_number, stripped[:100]))

    if any(path_fragment in line for path_fragment in (
        "/content/",
        "/kaggle/",
        "/home/",
    )):
        absolute_paths.append((line_number, line.strip()[:120]))

    for secret_type, pattern in secret_patterns.items():
        if pattern.search(line):
            possible_secret_locations.append(
                (line_number, secret_type)
            )


# ---------------------------------------------------------
# 4. Save extracted preview outside repository
# ---------------------------------------------------------

preview_path = INCOMING_DIR / "extracted_source_preview.py"
preview_path.write_text(extracted_code, encoding="utf-8")


# ---------------------------------------------------------
# 5. Print report
# ---------------------------------------------------------

print("\n" + "=" * 72)
print("1. SOURCE SUMMARY")
print("=" * 72)

print(f"Source type: {suffix}")
print(f"Extracted code size: {len(extracted_code.encode('utf-8')):,} bytes")
print(f"Extracted code lines: {len(code_lines):,}")

if notebook_info:
    print(f"Total notebook cells: {notebook_info['total_cells']:,}")
    print(f"Code cells: {notebook_info['code_cells']:,}")
    print(f"Markdown cells: {notebook_info['markdown_cells']:,}")
    print(f"Raw cells: {notebook_info['raw_cells']:,}")
    print(f"Stored outputs: {notebook_info['stored_outputs']:,}")


print("\n" + "=" * 72)
print("2. COLAB/JUPYTER DIRECTIVES")
print("=" * 72)

if colab_directives:
    print(f"Found {len(colab_directives)} directive line(s):")

    for line_number, preview in colab_directives[:30]:
        print(f"• Line {line_number}: {preview}")

    if len(colab_directives) > 30:
        print(f"... and {len(colab_directives) - 30} more")
else:
    print("No !command or %magic directives found.")


print("\n" + "=" * 72)
print("3. ENVIRONMENT-SPECIFIC PATHS")
print("=" * 72)

if absolute_paths:
    print(f"Found {len(absolute_paths)} environment-specific path(s):")

    for line_number, preview in absolute_paths[:30]:
        print(f"• Line {line_number}: {preview}")

    if len(absolute_paths) > 30:
        print(f"... and {len(absolute_paths) - 30} more")
else:
    print("No obvious environment-specific paths found.")


print("\n" + "=" * 72)
print("4. SECRET SAFETY SCAN")
print("=" * 72)

if possible_secret_locations:
    print("⚠️ Possible secret(s) detected.")
    print("مقادیر محرمانه عمداً نمایش داده نمی‌شوند.")

    for line_number, secret_type in possible_secret_locations:
        print(f"• Line {line_number}: possible {secret_type}")
else:
    print("✅ No obvious hard-coded token or private key detected.")


print("\n" + "=" * 72)
print("5. TEMPORARY EXTRACTED FILE")
print("=" * 72)

print(preview_path)
print("\nاین فایل هنوز داخل Git Repository قرار نگرفته است.")

print("\n✅ دریافت و بررسی اولیه کد اصلی کامل شد.")
print("⛔ فعلاً git add، commit یا push انجام ندهید.")
````

````python
from pathlib import Path
import ast
import json
from collections import Counter, defaultdict

SOURCE_FILE = Path(
    "/content/incoming_source/extracted_source_preview.py"
)

REPORT_FILE = Path(
    "/content/incoming_source/"
    "extracted_source_structure_report.json"
)

if not SOURCE_FILE.exists():
    raise FileNotFoundError(
        f"Source file not found:\n{SOURCE_FILE}"
    )

source_text = SOURCE_FILE.read_text(
    encoding="utf-8",
    errors="replace",
)

tree = ast.parse(
    source_text,
    filename=str(SOURCE_FILE),
)

print("=" * 88)
print("STRUCTURAL INSPECTION — COGNITIVE SELF-MODEL SOURCE")
print("=" * 88)
print(f"File       : {SOURCE_FILE}")
print(f"Total lines: {len(source_text.splitlines())}")
print(f"AST nodes  : {sum(1 for _ in ast.walk(tree))}")

# ------------------------------------------------------------------------------
# Helpers
# ------------------------------------------------------------------------------

def safe_unparse(node):
    """Return readable Python source for an AST node when possible."""
    try:
        return ast.unparse(node)
    except Exception:
        return f"<unparse unavailable: {type(node).__name__}>"

def decorators_of(node):
    return [
        safe_unparse(decorator)
        for decorator in getattr(node, "decorator_list", [])
    ]

def function_record(node, parent_class=None):
    return {
        "name": node.name,
        "line": node.lineno,
        "end_line": getattr(node, "end_lineno", node.lineno),
        "kind": (
            "async_function"
            if isinstance(node, ast.AsyncFunctionDef)
            else "function"
        ),
        "parent_class": parent_class,
        "decorators": decorators_of(node),
        "arguments": safe_unparse(node.args),
        "returns": (
            safe_unparse(node.returns)
            if node.returns is not None
            else None
        ),
        "docstring": ast.get_docstring(node),
    }

def class_record(node):
    methods = []

    for item in node.body:
        if isinstance(
            item,
            (ast.FunctionDef, ast.AsyncFunctionDef),
        ):
            methods.append(
                function_record(
                    item,
                    parent_class=node.name,
                )
            )

    return {
        "name": node.name,
        "line": node.lineno,
        "end_line": getattr(node, "end_lineno", node.lineno),
        "decorators": decorators_of(node),
        "bases": [
            safe_unparse(base)
            for base in node.bases
        ],
        "keywords": [
            safe_unparse(keyword)
            for keyword in node.keywords
        ],
        "docstring": ast.get_docstring(node),
        "method_count": len(methods),
        "methods": methods,
    }

def is_main_guard(node):
    """
    Detect:
        if __name__ == "__main__":
    """
    if not isinstance(node, ast.If):
        return False

    test = node.test

    return (
        isinstance(test, ast.Compare)
        and isinstance(test.left, ast.Name)
        and test.left.id == "__name__"
        and len(test.ops) == 1
        and isinstance(test.ops[0], ast.Eq)
        and len(test.comparators) == 1
        and isinstance(test.comparators[0], ast.Constant)
        and test.comparators[0].value == "__main__"
    )

# ------------------------------------------------------------------------------
# Collect imports
# ------------------------------------------------------------------------------

imports = []

for node in ast.walk(tree):
    if isinstance(node, ast.Import):
        for alias in node.names:
            imports.append(
                {
                    "line": node.lineno,
                    "type": "import",
                    "module": alias.name,
                    "alias": alias.asname,
                }
            )

    elif isinstance(node, ast.ImportFrom):
        module_name = node.module or ""

        for alias in node.names:
            imports.append(
                {
                    "line": node.lineno,
                    "type": "from_import",
                    "module": module_name,
                    "name": alias.name,
                    "alias": alias.asname,
                    "relative_level": node.level,
                }
            )

imports.sort(
    key=lambda item: (
        item["line"],
        item["module"],
    )
)

# ------------------------------------------------------------------------------
# Collect top-level definitions and execution blocks
# ------------------------------------------------------------------------------

top_level_functions = []
top_level_classes = []
main_guards = []
top_level_assignments = []
top_level_calls = []
top_level_other_statements = []

for node in tree.body:
    if isinstance(
        node,
        (ast.FunctionDef, ast.AsyncFunctionDef),
    ):
        top_level_functions.append(
            function_record(node)
        )

    elif isinstance(node, ast.ClassDef):
        top_level_classes.append(
            class_record(node)
        )

    elif is_main_guard(node):
        main_guards.append(
            {
                "line": node.lineno,
                "end_line": getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),
                "condition": safe_unparse(node.test),
                "body_statements": [
                    {
                        "line": statement.lineno,
                        "type": type(statement).__name__,
                        "source": safe_unparse(statement),
                    }
                    for statement in node.body
                ],
            }
        )

    elif isinstance(
        node,
        (ast.Assign, ast.AnnAssign, ast.AugAssign),
    ):
        top_level_assignments.append(
            {
                "line": node.lineno,
                "type": type(node).__name__,
                "source": safe_unparse(node),
            }
        )

    elif isinstance(node, ast.Expr):
        top_level_calls.append(
            {
                "line": node.lineno,
                "source": safe_unparse(node.value),
            }
        )

    elif not isinstance(
        node,
        (ast.Import, ast.ImportFrom),
    ):
        top_level_other_statements.append(
            {
                "line": node.lineno,
                "type": type(node).__name__,
                "source": safe_unparse(node),
            }
        )

# ------------------------------------------------------------------------------
# Detect possible duplicate top-level names
# ------------------------------------------------------------------------------

all_top_level_names = (
    [item["name"] for item in top_level_functions]
    + [item["name"] for item in top_level_classes]
)

name_counts = Counter(all_top_level_names)

duplicate_top_level_names = sorted(
    [
        name
        for name, count in name_counts.items()
        if count > 1
    ]
)

# ------------------------------------------------------------------------------
# Detect dataclasses and likely configuration objects
# ------------------------------------------------------------------------------

dataclasses = [
    item
    for item in top_level_classes
    if any(
        "dataclass" in decorator
        for decorator in item["decorators"]
    )
]

configuration_like_assignments = [
    item
    for item in top_level_assignments
    if any(
        keyword in item["source"].lower()
        for keyword in [
            "config",
            "cfg",
            "setting",
            "parameter",
            "seed",
            "device",
        ]
    )
]

# ------------------------------------------------------------------------------
# Create persistent report
# ------------------------------------------------------------------------------

report = {
    "source_file": str(SOURCE_FILE),
    "source_line_count": len(source_text.splitlines()),
    "ast_node_count": sum(1 for _ in ast.walk(tree)),
    "summary": {
        "import_entries": len(imports),
        "top_level_functions": len(top_level_functions),
        "top_level_classes": len(top_level_classes),
        "dataclasses": len(dataclasses),
        "main_guards": len(main_guards),
        "top_level_assignments": len(top_level_assignments),
        "top_level_expression_calls": len(top_level_calls),
        "duplicate_top_level_names": duplicate_top_level_names,
    },
    "imports": imports,
    "top_level_functions": top_level_functions,
    "top_level_classes": top_level_classes,
    "dataclasses": dataclasses,
    "main_guards": main_guards,
    "top_level_assignments": top_level_assignments,
    "configuration_like_assignments": (
        configuration_like_assignments
    ),
    "top_level_expression_calls": top_level_calls,
    "top_level_other_statements": (
        top_level_other_statements
    ),
}

REPORT_FILE.write_text(
    json.dumps(
        report,
        indent=2,
        ensure_ascii=False,
        default=str,
    ),
    encoding="utf-8",
)

# ------------------------------------------------------------------------------
# Concise console summary
# ------------------------------------------------------------------------------

print("\n" + "-" * 88)
print("SUMMARY")
print("-" * 88)

for key, value in report["summary"].items():
    print(f"{key:>34}: {value}")

print("\n" + "-" * 88)
print("TOP-LEVEL CLASSES")
print("-" * 88)

if top_level_classes:
    for item in top_level_classes:
        base_text = (
            f"({', '.join(item['bases'])})"
            if item["bases"]
            else ""
        )
        print(
            f"{item['line']:>6} | "
            f"class {item['name']}{base_text} "
            f"[methods={item['method_count']}]"
        )
else:
    print("No top-level classes found.")

print("\n" + "-" * 88)
print("TOP-LEVEL FUNCTIONS")
print("-" * 88)

if top_level_functions:
    for item in top_level_functions:
        print(
            f"{item['line']:>6} | "
            f"{item['kind']} {item['name']}("
            f"{item['arguments']})"
        )
else:
    print("No top-level functions found.")

print("\n" + "-" * 88)
print("MAIN GUARDS")
print("-" * 88)

if main_guards:
    for guard in main_guards:
        print(
            f"{guard['line']:>6} | "
            f"{guard['condition']}"
        )

        for statement in guard["body_statements"]:
            print(
                f"         └─ line {statement['line']}: "
                f"{statement['source']}"
            )
else:
    print("No if __name__ == '__main__' guard found.")

print("\n" + "-" * 88)
print("TOP-LEVEL EXPRESSION CALLS (POSSIBLE IMPORT-TIME EXECUTION)")
print("-" * 88)

if top_level_calls:
    for item in top_level_calls:
        print(
            f"{item['line']:>6} | "
            f"{item['source']}"
        )
else:
    print("✅ No standalone top-level expression calls found.")

print("\n" + "-" * 88)
print("REPORT SAVED")
print("-" * 88)
print(f"✅ {REPORT_FILE}")
````

````python
from pathlib import Path
import ast
import json

SOURCE_FILE = Path(
    "/content/incoming_source/extracted_source_preview.py"
)

TIMELINE_REPORT_FILE = Path(
    "/content/incoming_source/"
    "extracted_source_execution_timeline.json"
)

source_text = SOURCE_FILE.read_text(
    encoding="utf-8",
    errors="replace",
)

source_lines = source_text.splitlines()

tree = ast.parse(
    source_text,
    filename=str(SOURCE_FILE),
)

print("=" * 88)
print("TOP-LEVEL EXECUTION TIMELINE")
print("=" * 88)
print(f"Source : {SOURCE_FILE}")
print(f"Lines  : {len(source_lines)}")

def safe_unparse(node):
    try:
        return ast.unparse(node)
    except Exception:
        return f"<{type(node).__name__}>"

def compact_preview(node, max_length=180):
    text = safe_unparse(node)
    text = " ".join(text.split())

    if len(text) > max_length:
        text = text[:max_length - 3] + "..."

    return text

def classify_top_level_node(node):
    if isinstance(node, ast.Import):
        return "IMPORT"

    if isinstance(node, ast.ImportFrom):
        return "IMPORT_FROM"

    if isinstance(node, ast.ClassDef):
        return "CLASS_DEFINITION"

    if isinstance(node, ast.FunctionDef):
        return "FUNCTION_DEFINITION"

    if isinstance(node, ast.AsyncFunctionDef):
        return "ASYNC_FUNCTION_DEFINITION"

    if isinstance(node, (ast.Assign, ast.AnnAssign, ast.AugAssign)):
        return "ASSIGNMENT"

    if isinstance(node, ast.Expr):
        # نخستین Expr ممکن است docstring ماژول باشد.
        if (
            isinstance(node.value, ast.Constant)
            and isinstance(node.value.value, str)
        ):
            return "MODULE_DOCSTRING"

        return "EXPRESSION_OR_CALL"

    if isinstance(node, ast.If):
        return "IF_BLOCK"

    if isinstance(node, ast.For):
        return "FOR_BLOCK"

    if isinstance(node, ast.While):
        return "WHILE_BLOCK"

    if isinstance(node, ast.Try):
        return "TRY_BLOCK"

    if isinstance(node, ast.With):
        return "WITH_BLOCK"

    return type(node).__name__.upper()

def is_definition_or_import(category):
    return category in {
        "IMPORT",
        "IMPORT_FROM",
        "CLASS_DEFINITION",
        "FUNCTION_DEFINITION",
        "ASYNC_FUNCTION_DEFINITION",
        "MODULE_DOCSTRING",
    }

timeline = []

for index, node in enumerate(tree.body, start=1):
    category = classify_top_level_node(node)

    timeline.append(
        {
            "index": index,
            "category": category,
            "line": node.lineno,
            "end_line": getattr(node, "end_lineno", node.lineno),
            "preview": compact_preview(node),
        }
    )

# ------------------------------------------------------------------------------
# Detect contiguous executable islands:
# Anything that is not import / definition / module docstring.
# ------------------------------------------------------------------------------

execution_islands = []
current_island = []

for item in timeline:
    if is_definition_or_import(item["category"]):
        if current_island:
            execution_islands.append(current_island)
            current_island = []
    else:
        current_island.append(item)

if current_island:
    execution_islands.append(current_island)

island_report = []

for island_number, island in enumerate(
    execution_islands,
    start=1,
):
    categories = sorted(
        set(item["category"] for item in island)
    )

    island_report.append(
        {
            "island_number": island_number,
            "start_line": island[0]["line"],
            "end_line": island[-1]["end_line"],
            "statement_count": len(island),
            "categories": categories,
            "statements": island,
        }
    )

report = {
    "source_file": str(SOURCE_FILE),
    "total_lines": len(source_lines),
    "top_level_statement_count": len(timeline),
    "top_level_timeline": timeline,
    "execution_island_count": len(island_report),
    "execution_islands": island_report,
}

TIMELINE_REPORT_FILE.write_text(
    json.dumps(
        report,
        indent=2,
        ensure_ascii=False,
    ),
    encoding="utf-8",
)

print("\n" + "-" * 88)
print("EXECUTION ISLAND SUMMARY")
print("-" * 88)

for island in island_report:
    category_text = ", ".join(island["categories"])

    print(
        f"\nIsland {island['island_number']:>2} | "
        f"lines {island['start_line']}-{island['end_line']} | "
        f"statements={island['statement_count']}"
    )
    print(f"Categories: {category_text}")

    for statement in island["statements"][:4]:
        print(
            f"  {statement['line']:>6}-{statement['end_line']:<6} | "
            f"{statement['category']:<22} | "
            f"{statement['preview']}"
        )

    if len(island["statements"]) > 4:
        remaining = len(island["statements"]) - 4
        print(f"  ... ({remaining} additional statements)")

print("\n" + "-" * 88)
print("HIGH-RISK EXECUTION REGION")
print("-" * 88)

high_risk_items = [
    item
    for item in timeline
    if item["line"] >= 1500
    and not is_definition_or_import(item["category"])
]

if not high_risk_items:
    print("No execution statements detected after line 1500.")
else:
    for item in high_risk_items:
        print(
            f"{item['line']:>6}-{item['end_line']:<6} | "
            f"{item['category']:<22} | "
            f"{item['preview']}"
        )

print("\n" + "-" * 88)
print("REPORT SAVED")
print("-" * 88)
print(f"✅ {TIMELINE_REPORT_FILE}")
````

````python
from pathlib import Path
import ast
import importlib.util
import sys
import traceback

SOURCE_FILE = Path(
    "/content/incoming_source/extracted_source_preview.py"
)

SAFE_OUTPUT_FILE = Path(
    "/content/incoming_source/"
    "cognitive_self_model_m20_6_3_2_safe.py"
)

RUNNER_NAME = "run_m20_6_3_2_experiment"

if not SOURCE_FILE.exists():
    raise FileNotFoundError(
        f"Source file was not found:\n{SOURCE_FILE}"
    )

source_text = SOURCE_FILE.read_text(
    encoding="utf-8",
    errors="replace",
)

original_tree = ast.parse(
    source_text,
    filename=str(SOURCE_FILE),
)

print("=" * 88)
print("SAFE MODULE CONVERSION — M20.6.3.2")
print("=" * 88)
print(f"Source : {SOURCE_FILE}")
print(f"Output : {SAFE_OUTPUT_FILE}")


# ------------------------------------------------------------------------------
# A module-level definition or import must remain directly importable.
# Everything else is runtime experiment logic and goes inside the runner.
# ------------------------------------------------------------------------------

def is_module_definition_or_import(node):
    return isinstance(
        node,
        (
            ast.Import,
            ast.ImportFrom,
            ast.ClassDef,
            ast.FunctionDef,
            ast.AsyncFunctionDef,
        ),
    )


def is_module_docstring(node):
    return (
        isinstance(node, ast.Expr)
        and isinstance(node.value, ast.Constant)
        and isinstance(node.value.value, str)
    )


# ------------------------------------------------------------------------------
# Gather names assigned by runtime statements.
#
# They need to remain module globals because many reusable helper functions
# reference them through Python's normal module-global resolution.
# ------------------------------------------------------------------------------

class RuntimeAssignedNameCollector(ast.NodeVisitor):
    def __init__(self):
        self.names = set()

    def visit_Name(self, node):
        if isinstance(node.ctx, ast.Store):
            self.names.add(node.id)

    def visit_FunctionDef(self, node):
        # A nested function creates a separate scope.
        return

    def visit_AsyncFunctionDef(self, node):
        return

    def visit_Lambda(self, node):
        return

    def visit_ClassDef(self, node):
        return


# ------------------------------------------------------------------------------
# Python does not allow a global declaration together with an annotated
# assignment inside a function:
#
#     global GROUPS
#     GROUPS: Dict[...] = ...
#
# Therefore, runtime AnnAssign nodes become standard assignments.
# The runtime value and scientific behavior remain unchanged.
# ------------------------------------------------------------------------------

class RuntimeAnnotationNormalizer(ast.NodeTransformer):
    def visit_AnnAssign(self, node):
        self.generic_visit(node)

        # Annotation with a runtime value:
        #     NAME: Type = value
        # becomes:
        #     NAME = value
        if node.value is not None:
            return ast.copy_location(
                ast.Assign(
                    targets=[node.target],
                    value=node.value,
                ),
                node,
            )

        # Pure runtime annotation:
        #     NAME: Type
        # has no needed runtime effect for this experiment.
        return ast.copy_location(
            ast.Pass(),
            node,
        )


module_body = []
runtime_nodes = []

for node in original_tree.body:
    if is_module_docstring(node):
        module_body.append(node)

    elif is_module_definition_or_import(node):
        module_body.append(node)

    else:
        runtime_nodes.append(node)

if not runtime_nodes:
    raise RuntimeError(
        "No runtime statements were found. "
        "The source does not need this conversion."
    )

# Normalize annotated runtime assignments before placing them in a function.
normalizer = RuntimeAnnotationNormalizer()
runtime_nodes = [
    normalizer.visit(node)
    for node in runtime_nodes
]

runtime_nodes = [
    node
    for node in runtime_nodes
    if node is not None
]

for node in runtime_nodes:
    ast.fix_missing_locations(node)

collector = RuntimeAssignedNameCollector()

for node in runtime_nodes:
    collector.visit(node)

runtime_global_names = sorted(collector.names)

if not runtime_global_names:
    raise RuntimeError(
        "No runtime-assigned global names were detected."
    )

# The `global` declaration must be first in the runner body.
global_declaration = ast.Global(
    names=runtime_global_names
)

runner_docstring = ast.Expr(
    value=ast.Constant(
        value=(
            "Execute the complete M20.6.3.2 stable-mechanism "
            "structural-oracle experiment explicitly."
        )
    )
)

runner_function = ast.FunctionDef(
    name=RUNNER_NAME,
    args=ast.arguments(
        posonlyargs=[],
        args=[],
        kwonlyargs=[],
        kw_defaults=[],
        defaults=[],
    ),
    body=[
        runner_docstring,
        global_declaration,
        *runtime_nodes,
    ],
    decorator_list=[],
    returns=None,
    type_comment=None,
)

# Permit explicit command-line / direct-script execution,
# while keeping ordinary imports safe.
main_guard = ast.If(
    test=ast.Compare(
        left=ast.Name(
            id="__name__",
            ctx=ast.Load(),
        ),
        ops=[ast.Eq()],
        comparators=[
            ast.Constant(
                value="__main__"
            )
        ],
    ),
    body=[
        ast.Expr(
            value=ast.Call(
                func=ast.Name(
                    id=RUNNER_NAME,
                    ctx=ast.Load(),
                ),
                args=[],
                keywords=[],
            )
        )
    ],
    orelse=[],
)

safe_tree = ast.Module(
    body=[
        *module_body,
        runner_function,
        main_guard,
    ],
    type_ignores=[],
)

safe_tree = ast.fix_missing_locations(safe_tree)

safe_source = ast.unparse(safe_tree) + "\n"

SAFE_OUTPUT_FILE.write_text(
    safe_source,
    encoding="utf-8",
)

print("\n" + "-" * 88)
print("CONVERSION SUMMARY")
print("-" * 88)
print(f"Module-level retained nodes : {len(module_body)}")
print(f"Runtime nodes moved         : {len(runtime_nodes)}")
print(f"Runtime global names        : {len(runtime_global_names)}")
print(f"Runner function             : {RUNNER_NAME}")
print(f"Safe file written           : {SAFE_OUTPUT_FILE}")

print("\nFirst 30 runtime globals:")
for name in runtime_global_names[:30]:
    print(f"  - {name}")

if len(runtime_global_names) > 30:
    print(
        f"  ... plus "
        f"{len(runtime_global_names) - 30} additional globals"
    )

# ------------------------------------------------------------------------------
# 1. Syntax validation
# ------------------------------------------------------------------------------

print("\n" + "-" * 88)
print("SYNTAX VALIDATION")
print("-" * 88)

compile(
    safe_source,
    str(SAFE_OUTPUT_FILE),
    "exec",
)

print("✅ Safe module compiles successfully.")

# ------------------------------------------------------------------------------
# 2. Structural validation:
#    No top-level runtime expressions / assignments should remain,
#    except imports, definitions, module docstring, and main guard.
# ------------------------------------------------------------------------------

safe_tree_check = ast.parse(
    safe_source,
    filename=str(SAFE_OUTPUT_FILE),
)

unexpected_top_level_nodes = []

for node in safe_tree_check.body:
    allowed = (
        is_module_docstring(node)
        or is_module_definition_or_import(node)
        or isinstance(node, ast.If)
    )

    if not allowed:
        unexpected_top_level_nodes.append(
            {
                "line": node.lineno,
                "type": type(node).__name__,
                "source": ast.unparse(node)[:300],
            }
        )

runner_nodes = [
    node
    for node in safe_tree_check.body
    if isinstance(node, ast.FunctionDef)
    and node.name == RUNNER_NAME
]

main_guards = [
    node
    for node in safe_tree_check.body
    if isinstance(node, ast.If)
]

print(
    f"Runner definitions found   : {len(runner_nodes)}"
)
print(
    f"Top-level main guards found: {len(main_guards)}"
)
print(
    f"Unexpected top-level nodes : "
    f"{len(unexpected_top_level_nodes)}"
)

if len(runner_nodes) != 1:
    raise RuntimeError(
        "Expected exactly one explicit experiment runner."
    )

if unexpected_top_level_nodes:
    raise RuntimeError(
        "Unsafe top-level runtime nodes remain:\n"
        f"{unexpected_top_level_nodes}"
    )

print(
    "✅ No unintended executable experiment statements remain "
    "at module import level."
)

# ------------------------------------------------------------------------------
# 3. Safe import test:
#    This imports definitions only. It must NOT instantiate CFG,
#    generate groups, load the LLM, fit models, or write telemetry.
# ------------------------------------------------------------------------------

print("\n" + "-" * 88)
print("SAFE IMPORT TEST — NO EXPERIMENT EXECUTION")
print("-" * 88)

test_module_name = (
    "_m20_6_3_2_safe_import_test"
)

sys.modules.pop(
    test_module_name,
    None,
)

module_spec = importlib.util.spec_from_file_location(
    test_module_name,
    SAFE_OUTPUT_FILE,
)

if module_spec is None or module_spec.loader is None:
    raise ImportError(
        "Could not construct an import specification "
        "for the safe module."
    )

safe_module = importlib.util.module_from_spec(
    module_spec
)

sys.modules[test_module_name] = safe_module

try:
    module_spec.loader.exec_module(
        safe_module
    )
except Exception:
    print(
        "❌ Safe import test failed. "
        "This is an import/dependency issue, not an experiment run."
    )
    traceback.print_exc()
    raise
finally:
    sys.modules.pop(
        test_module_name,
        None,
    )

if not hasattr(
    safe_module,
    RUNNER_NAME,
):
    raise RuntimeError(
        f"Runner `{RUNNER_NAME}` was not importable."
    )

# CFG was originally created at top level. Its absence after import proves
# that experiment runtime code has not executed.
if hasattr(safe_module, "CFG"):
    raise RuntimeError(
        "Unsafe import detected: CFG exists after import, "
        "which means runtime code executed unexpectedly."
    )

# Model should not exist after a safe import either.
if hasattr(safe_module, "model"):
    raise RuntimeError(
        "Unsafe import detected: model exists after import, "
        "which means the Transformer was loaded unexpectedly."
    )

print("✅ Module imported safely.")
print("✅ CFG was not instantiated during import.")
print("✅ Transformer model was not loaded during import.")
print("✅ No experimental pipeline was automatically executed.")

print("\n" + "=" * 88)
print("SAFE MODULE CONVERSION COMPLETED")
print("=" * 88)
print("\nSafe import usage:")
print(
    "  from cognitive_self_model_m20_6_3_2_safe "
    f"import {RUNNER_NAME}"
)
print()
print("Explicit experiment execution:")
print(f"  {RUNNER_NAME}()")
````

````python
from pathlib import Path
import ast
import importlib.util
import sys
import traceback

SOURCE_SAFE_FILE = Path(
    "/content/incoming_source/"
    "cognitive_self_model_m20_6_3_2_safe.py"
)

IMPORT_SAFE_FILE = Path(
    "/content/incoming_source/"
    "cognitive_self_model_m20_6_3_2_import_safe.py"
)

RUNNER_NAME = "run_m20_6_3_2_experiment"

if not SOURCE_SAFE_FILE.exists():
    raise FileNotFoundError(
        f"Safe source file was not found:\n{SOURCE_SAFE_FILE}"
    )

source_text = SOURCE_SAFE_FILE.read_text(
    encoding="utf-8",
    errors="replace",
)

tree = ast.parse(
    source_text,
    filename=str(SOURCE_SAFE_FILE),
)

print("=" * 88)
print("CORRECTED LAZY DEPENDENCY PATCH — TRANSFORMERLENS")
print("=" * 88)
print(f"Input  : {SOURCE_SAFE_FILE}")
print(f"Output : {IMPORT_SAFE_FILE}")


def is_transformer_lens_import(node):
    return (
        isinstance(node, ast.ImportFrom)
        and node.module == "transformer_lens"
        and any(
            alias.name == "HookedTransformer"
            for alias in node.names
        )
    )


# ==============================================================================
# 1. Remove TransformerLens only from MODULE scope.
# ==============================================================================

original_module_body = list(tree.body)

filtered_module_body = [
    node
    for node in original_module_body
    if not is_transformer_lens_import(node)
]

removed_count = (
    len(original_module_body)
    - len(filtered_module_body)
)

if removed_count != 1:
    raise RuntimeError(
        "Expected exactly one module-level TransformerLens import, "
        f"but removed {removed_count}."
    )

# Critical correction: install filtered nodes back into the actual AST.
tree.body = filtered_module_body

print("\n✅ Removed exactly one module-level import:")
print(
    "   from transformer_lens import HookedTransformer"
)


# ==============================================================================
# 2. Locate the runner in the corrected AST.
# ==============================================================================

runner_candidates = [
    node
    for node in tree.body
    if isinstance(node, ast.FunctionDef)
    and node.name == RUNNER_NAME
]

if len(runner_candidates) != 1:
    raise RuntimeError(
        f"Expected exactly one runner named `{RUNNER_NAME}`, "
        f"found {len(runner_candidates)}."
    )

runner = runner_candidates[0]


# ==============================================================================
# 3. Confirm no previous lazy import exists in this runner.
# ==============================================================================

existing_runner_imports = [
    node
    for node in ast.walk(runner)
    if is_transformer_lens_import(node)
]

if existing_runner_imports:
    raise RuntimeError(
        "The source safe file unexpectedly already contains "
        "a TransformerLens import inside the runner. "
        "Use the original `_safe.py` file as input, not a previously "
        "patched file."
    )


# ==============================================================================
# 4. Find exact model-load statement and insert lazy import before it.
# ==============================================================================

model_load_index = None

for index, node in enumerate(runner.body):
    if not isinstance(node, ast.Assign):
        continue

    rendered = ast.unparse(node)

    is_model_assignment = any(
        isinstance(target, ast.Name)
        and target.id == "model"
        for target in node.targets
    )

    if (
        is_model_assignment
        and "HookedTransformer.from_pretrained" in rendered
    ):
        model_load_index = index
        break

if model_load_index is None:
    raise RuntimeError(
        "Could not locate:\n"
        "model = HookedTransformer.from_pretrained(...)"
    )

lazy_dependency_import = ast.ImportFrom(
    module="transformer_lens",
    names=[
        ast.alias(
            name="HookedTransformer",
            asname=None,
        )
    ],
    level=0,
)

lazy_dependency_guard = ast.Try(
    body=[
        lazy_dependency_import,
    ],
    handlers=[
        ast.ExceptHandler(
            type=ast.Name(
                id="ModuleNotFoundError",
                ctx=ast.Load(),
            ),
            name="error",
            body=[
                ast.Raise(
                    exc=ast.Call(
                        func=ast.Name(
                            id="RuntimeError",
                            ctx=ast.Load(),
                        ),
                        args=[
                            ast.Constant(
                                value=(
                                    "The explicit M20.6.3.2 experiment requires "
                                    "`transformer_lens`, but it is not installed. "
                                    "Install it before calling "
                                    "`run_m20_6_3_2_experiment()`."
                                )
                            )
                        ],
                        keywords=[],
                    ),
                    cause=ast.Name(
                        id="error",
                        ctx=ast.Load(),
                    ),
                )
            ],
        )
    ],
    orelse=[],
    finalbody=[],
)

runner.body.insert(
    model_load_index,
    lazy_dependency_guard,
)

tree = ast.fix_missing_locations(tree)

patched_source = ast.unparse(tree) + "\n"

IMPORT_SAFE_FILE.write_text(
    patched_source,
    encoding="utf-8",
)

print("\n" + "-" * 88)
print("PATCH SUMMARY")
print("-" * 88)
print("✅ Removed TransformerLens import from module scope.")
print("✅ Added one lazy import inside the explicit runner.")
print(
    "✅ Lazy import location: immediately before "
    "`HookedTransformer.from_pretrained(...)`."
)
print(f"✅ Patched file written: {IMPORT_SAFE_FILE}")


# ==============================================================================
# 5. Syntax validation.
# ==============================================================================

print("\n" + "-" * 88)
print("SYNTAX VALIDATION")
print("-" * 88)

compile(
    patched_source,
    str(IMPORT_SAFE_FILE),
    "exec",
)

print("✅ Import-safe module compiles successfully.")


# ==============================================================================
# 6. Strict AST-level placement verification.
# ==============================================================================

patched_tree = ast.parse(
    patched_source,
    filename=str(IMPORT_SAFE_FILE),
)

top_level_transformer_lens_imports = [
    node
    for node in patched_tree.body
    if is_transformer_lens_import(node)
]

patched_runner_candidates = [
    node
    for node in patched_tree.body
    if isinstance(node, ast.FunctionDef)
    and node.name == RUNNER_NAME
]

if len(patched_runner_candidates) != 1:
    raise RuntimeError(
        "The runner was not preserved correctly."
    )

patched_runner = patched_runner_candidates[0]

runner_transformer_lens_imports = [
    node
    for node in ast.walk(patched_runner)
    if is_transformer_lens_import(node)
]

print(
    "Top-level TransformerLens imports : "
    f"{len(top_level_transformer_lens_imports)}"
)
print(
    "Runner TransformerLens imports    : "
    f"{len(runner_transformer_lens_imports)}"
)

if len(top_level_transformer_lens_imports) != 0:
    raise RuntimeError(
        "FAIL: TransformerLens is still imported at module scope."
    )

if len(runner_transformer_lens_imports) != 1:
    raise RuntimeError(
        "FAIL: Expected exactly one TransformerLens import "
        "inside the runner."
    )

print(
    "✅ Dependency placement is correct: "
    "0 top-level / 1 runner-level import."
)


# ==============================================================================
# 7. Actual import-safe test.
#
# transformer_lens is intentionally not installed in this environment.
# Import must still work, because the model dependency is now lazy.
# ==============================================================================

print("\n" + "-" * 88)
print("IMPORT-SAFE TEST — NO TRANSFORMERLENS REQUIRED")
print("-" * 88)

test_module_name = "_m20_6_3_2_import_safe_test"

sys.modules.pop(test_module_name, None)

module_spec = importlib.util.spec_from_file_location(
    test_module_name,
    IMPORT_SAFE_FILE,
)

if module_spec is None or module_spec.loader is None:
    raise ImportError(
        "Could not create import specification "
        "for the patched module."
    )

safe_module = importlib.util.module_from_spec(
    module_spec
)

sys.modules[test_module_name] = safe_module

try:
    module_spec.loader.exec_module(safe_module)

except Exception:
    print("❌ Import-safe test failed unexpectedly.")
    traceback.print_exc()
    raise

finally:
    sys.modules.pop(test_module_name, None)


required_definitions = [
    "Config",
    "StreamRecord",
    "PlattCalibrator",
    "generate_group",
    "build_temporal_dataset",
    "structural_oracle",
    RUNNER_NAME,
]

missing_definitions = [
    name
    for name in required_definitions
    if not hasattr(safe_module, name)
]

if missing_definitions:
    raise RuntimeError(
        "Reusable definitions missing after import:\n"
        f"{missing_definitions}"
    )

unsafe_runtime_names = [
    name
    for name in (
        "CFG",
        "GROUPS",
        "RAW_DATASETS_BY_HORIZON",
        "DATASETS_BY_HORIZON",
        "DATASETS",
        "model",
        "telemetry",
    )
    if hasattr(safe_module, name)
]

if unsafe_runtime_names:
    raise RuntimeError(
        "Runtime code was executed during import:\n"
        f"{unsafe_runtime_names}"
    )

print("✅ Module imported without transformer_lens.")
print("✅ Reusable classes/functions are available.")
print("✅ CFG was not instantiated.")
print("✅ Data streams were not generated.")
print("✅ Transformer was not loaded.")
print("✅ No experiment, fitting, or telemetry execution occurred.")

print("\n" + "=" * 88)
print("IMPORT-SAFE MODULE COMPLETED")
print("=" * 88)

print("\nValidated module:")
print(f"  {IMPORT_SAFE_FILE}")

print("\nSafe import:")
print(
    "  from cognitive_self_model_m20_6_3_2_import_safe "
    f"import {RUNNER_NAME}"
)

print("\nExplicit execution after dependency installation:")
print(f"  {RUNNER_NAME}()")
````

````python
# ================================================================
# EMERGENCY PROJECT BACKUP — SAFE / READ-ONLY
# همهٔ فایل‌های قابل‌ارتباط Runtime را در Google Drive ذخیره می‌کند.
# هیچ فایل اصلی حذف یا جابه‌جا نمی‌شود.
# ================================================================

from google.colab import drive
from pathlib import Path
from datetime import datetime
import os
import shutil
import zipfile

# اتصال به Google Drive
drive.mount("/content/drive", force_remount=False)

TIMESTAMP = datetime.now().strftime("%Y%m%d_%H%M%S")

DRIVE_BACKUP_ROOT = Path("/content/drive/MyDrive/cognitive_self_model_backups")
BACKUP_DIR = DRIVE_BACKUP_ROOT / f"runtime_backup_{TIMESTAMP}"
BACKUP_DIR.mkdir(parents=True, exist_ok=True)

# مسیرهای محتملی که ممکن است فایل‌های پروژه در آن‌ها باشند
CANDIDATE_SOURCES = [
    Path("/content/cognitive-self-model_pre_audit_backup"),
    Path("/content/cognitive-self-model"),
    Path("/content"),
]

# پوشه‌های حجیم/سیستمی که نباید بی‌دلیل backup شوند
SKIP_DIRS = {
    ".git", ".cache", ".config", ".local", "__pycache__",
    "sample_data", "node_modules",
}

# پسوندهای ارزشمند برای کد، Notebook، پیکربندی، گزارش و نتایج سبک
KEEP_SUFFIXES = {
    ".py", ".ipynb", ".md", ".txt",
    ".json", ".yaml", ".yml", ".toml",
    ".csv", ".tsv",
}

# کلیدواژه‌های پروژه؛ برای اسکن عمومی /content فقط فایل‌های مرتبط‌تر را می‌گیریم
KEYWORDS = (
    "cognitive", "self_model", "self-model",
    "milestone", "temporal", "epistemic",
    "governance", "activation", "causal",
    "forecast", "untitled",
)

MAX_FILE_SIZE_MB = 100
copied = []
skipped_large = []
seen_sources = set()

def is_relevant(file_path: Path, source_root: Path) -> bool:
    """برای پوشه‌های مشخص پروژه همهٔ فایل‌های کد را می‌گیریم؛
       برای /content عمومی فقط نام‌های دارای کلیدواژه را."""
    try:
        relative = str(file_path.relative_to(source_root)).lower()
    except ValueError:
        relative = file_path.name.lower()

    # مسیرهای مشخص پروژه: تمام فایل‌های مفید
    if source_root.name in {
        "cognitive-self-model",
        "cognitive-self-model_pre_audit_backup",
    }:
        return True

    # /content عمومی: فقط موارد مرتبط با پروژه
    return any(keyword in relative for keyword in KEYWORDS)

def copy_relevant_files(source_root: Path):
    if not source_root.exists():
        print(f"[NOT FOUND] {source_root}")
        return

    source_key = str(source_root.resolve())
    if source_key in seen_sources:
        return
    seen_sources.add(source_key)

    print(f"\nScanning: {source_root}")

    for current, dirs, files in os.walk(source_root):
        current_path = Path(current)

        # از ورود به پوشه‌های غیرضروری جلوگیری می‌کنیم
        dirs[:] = [d for d in dirs if d not in SKIP_DIRS]

        for filename in files:
            src = current_path / filename

            if src.suffix.lower() not in KEEP_SUFFIXES:
                continue

            if not is_relevant(src, source_root):
                continue

            try:
                size_mb = src.stat().st_size / (1024 * 1024)
            except OSError:
                continue

            if size_mb > MAX_FILE_SIZE_MB:
                skipped_large.append((str(src), size_mb))
                continue

            try:
                # هر root را جدا نگه می‌داریم تا فایل‌ها overwrite نشوند
                root_label = source_root.name.replace("/", "_")
                relative = src.relative_to(source_root)
                dst = BACKUP_DIR / root_label / relative
                dst.parent.mkdir(parents=True, exist_ok=True)

                if not dst.exists():
                    shutil.copy2(src, dst)
                    copied.append((str(src), str(dst), size_mb))
            except Exception as exc:
                print(f"  [COPY FAILED] {src}: {exc}")

for root in CANDIDATE_SOURCES:
    copy_relevant_files(root)

# ساخت manifest خوانا
manifest = BACKUP_DIR / "BACKUP_MANIFEST.txt"
with manifest.open("w", encoding="utf-8") as f:
    f.write("COGNITIVE SELF-MODEL — RUNTIME BACKUP MANIFEST\n")
    f.write(f"Created: {datetime.now().isoformat()}\n")
    f.write("=" * 90 + "\n\n")
    f.write(f"Copied files: {len(copied)}\n\n")

    for src, dst, size_mb in copied:
        f.write(f"SOURCE: {src}\n")
        f.write(f"BACKUP: {dst}\n")
        f.write(f"SIZE_MB: {size_mb:.3f}\n\n")

    if skipped_large:
        f.write("\nSKIPPED LARGE FILES (> 100 MB):\n")
        for src, size_mb in skipped_large:
            f.write(f"{size_mb:.2f} MB  {src}\n")

# ساخت ZIP از backup برای دانلود یا انتقال آسان
zip_base = str(DRIVE_BACKUP_ROOT / f"runtime_backup_{TIMESTAMP}")
zip_path = shutil.make_archive(zip_base, "zip", root_dir=BACKUP_DIR)

print("\n" + "=" * 90)
print("BACKUP COMPLETED SUCCESSFULLY")
print("=" * 90)
print(f"Backup folder:\n{BACKUP_DIR}")
print(f"\nZIP archive:\n{zip_path}")
print(f"\nCopied relevant files: {len(copied)}")
print(f"Large files skipped: {len(skipped_large)}")
print("\nImportant: the current Notebook itself must still be saved via:")
print("File → Save a copy in Drive  AND  File → Download → Download .ipynb")
````

````python
# ============================================================
# COMMIT + PUSH CURRENT REPOSITORY TO GITHUB
# مسیر Repository: /content/cognitive-self-model
# Token در فایل یا Git ذخیره نمی‌شود.
# ============================================================

from pathlib import Path
import subprocess
import os
import getpass
import tempfile
import stat
import shutil

REPO = Path("/content/cognitive-self-model")

def run(cmd, *, env=None, check=True):
    result = subprocess.run(
        cmd,
        cwd=str(REPO),
        env=env,
        text=True,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
    )
    print(result.stdout, end="")
    if check and result.returncode != 0:
        raise RuntimeError(f"Command failed ({result.returncode}): {' '.join(cmd)}")
    return result

if not (REPO / ".git").is_dir():
    raise FileNotFoundError(f"Git repository پیدا نشد: {REPO}")

# تنظیم origin، فقط اگر قبلاً وجود نداشته باشد
origin = run(["git", "remote", "get-url", "origin"], check=False).stdout.strip()

if not origin:
    origin = input(
        "GitHub repository URL را وارد کن (مثال: https://github.com/USERNAME/REPOSITORY.git):\n"
    ).strip()
    if not origin:
        raise ValueError("Repository URL وارد نشده است.")
    run(["git", "remote", "add", "origin", origin])

print(f"\nRemote origin: {origin}")

# جلوگیری از ورود فایل‌های موقت، مدل‌های حجیم و Secretها به Git
gitignore = REPO / ".gitignore"
ignore_rules = """
# Colab / Python cache
__pycache__/
*.py[cod]
.ipynb_checkpoints/
.cache/
.config/

# Secrets — NEVER COMMIT
.env
.env.*
*.key
*.pem
credentials.json
token.json

# Large model/runtime artifacts
*.pt
*.pth
*.ckpt
*.bin
*.safetensors

# Local backups
cognitive_self_model_backups/
"""
existing = gitignore.read_text(encoding="utf-8") if gitignore.exists() else ""
if "# Colab / Python cache" not in existing:
    with gitignore.open("a", encoding="utf-8") as f:
        f.write("\n" + ignore_rules)

# در صورت نداشتن هویت Git، برای همین Repository تنظیم می‌شود
name = run(["git", "config", "user.name"], check=False).stdout.strip()
email = run(["git", "config", "user.email"], check=False).stdout.strip()

if not name:
    run(["git", "config", "user.name", "Cognitive Self Model Developer"])
if not email:
    run(["git", "config", "user.email", "developer@users.noreply.github.com"])

# افزودن همهٔ فایل‌های مجاز پروژه
run(["git", "add", "-A"])

print("\n========== FILES TO COMMIT ==========")
run(["git", "status", "--short"])

# Commit، فقط اگر تغییر جدیدی وجود دارد
has_staged_changes = subprocess.run(
    ["git", "diff", "--cached", "--quiet"],
    cwd=str(REPO)
).returncode != 0

if has_staged_changes:
    run([
        "git", "commit", "-m",
        "Update Cognitive Self-Model implementation, notebooks, configs, and documentation"
    ])
else:
    print("\nNo new staged changes: pushing existing commits if needed.")

# نام branch اصلی
run(["git", "branch", "-M", "main"])

# GitHub Personal Access Token:
# Fine-grained token: Repository permissions -> Contents: Read and write
token = getpass.getpass(
    "\nGitHub Personal Access Token را وارد کن (نمایش داده نمی‌شود): "
).strip()

if not token:
    raise ValueError("Token وارد نشده است.")

# Token فقط موقتاً و فقط برای همین Push استفاده می‌شود؛ در remote/config ذخیره نمی‌شود.
askpass_path = Path(tempfile.mkstemp(prefix="github_askpass_", suffix=".sh")[1])
askpass_path.write_text(
    '#!/bin/sh\n'
    'case "$1" in\n'
    '  *Username*) echo "x-access-token" ;;\n'
    '  *) echo "$GITHUB_TOKEN" ;;\n'
    'esac\n',
    encoding="utf-8"
)
askpass_path.chmod(askpass_path.stat().st_mode | stat.S_IXUSR)

push_env = os.environ.copy()
push_env["GITHUB_TOKEN"] = token
push_env["GIT_ASKPASS"] = str(askpass_path)
push_env["GIT_TERMINAL_PROMPT"] = "0"

try:
    print("\n========== PUSHING TO GITHUB ==========")
    run(["git", "push", "-u", "origin", "main"], env=push_env)
finally:
    # پاک‌سازی Token و ابزار موقت احراز هویت
    push_env.pop("GITHUB_TOKEN", None)
    if askpass_path.exists():
        askpass_path.unlink()

print("\n" + "=" * 60)
print("SUCCESS: Repository committed and pushed to GitHub.")
print("=" * 60)
print(f"Repository: {origin}")
print("Branch: main")
````
